```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("data.csv")

date_columns = [col for col in df.columns if 'date' in col]
close_columns = [col for col in df.columns if 'close' in col]
```

```python
close = df[close_columns]
```

```python
close["date"] = df["date_afran"]
```

```python
close
```

```python
close['date'] = pd.to_datetime(close['date'], format='%Y%m%d')
close
```

```python
close.set_index(close.date, inplace= True)
```

```python
close.drop(columns=["date"], inplace=True)
```

```python
close
```

```python
#close.to_csv("closeDate.csv")
```

```python
log_returns = np.log(close/close.shift(1)).dropna()
# Rename columns from 'close_*' to 'ret_*'
log_returns.columns = [col.replace('close_', 'ret_') for col in log_returns.columns]
```

```python
log_returns.to_csv("logRetData.csv")
```

```python
simple_returns = (((close / close.shift(1)) - 1)*100).dropna()
# Rename columns from 'close_*' to 'ret_*'
simple_returns.columns = [col.replace('close_', 'ret_') for col in simple_returns.columns]
```

```python
simple_returns.to_csv("simpleRetData.csv")
```
